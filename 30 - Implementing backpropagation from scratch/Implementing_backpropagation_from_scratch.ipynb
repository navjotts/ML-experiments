# A simple Neural Network

For simplicity, let's start with a simple 3 layer Neural Network

![Simple Neural Net](https://www.dropbox.com/s/5phpmx50s3vrlmk/simple_NN.png?raw=1)

# Initialize

Let's create variables to store weights and biases for the above network, and initialize each with $0.5$.


Naming convention $(w, b)$ for the `(input_neuron_weight, input_neuron_bias)` 

and $(w_i, b_i)$ for `(hidden_neuron_weights, hidden_neuron_bias)` 

```
import numpy as np

b, w, b1, w1, b2, w2 = np.ones(6)*0.5
print("b =", b)
print("w =", w)
print("b1 =", b1)
print("w1 =", w1)
print("b2 =", b2)
print("w2 =", w2)
```

# Sample set

Let's say the training set is given as

NN(4) == 0

NN(0.5) == 1

NN(0.125) == 1


```
x1, y1 = (4, 0)
x2, y2 = (0.5, 1)
x3, y3 = (0.125, 1)
```

### Activation function = ReLu

```
def activation(x):
  return max(0, x)
```

### Forward Propagation once

```
def feed_forward(x):
  h1 = activation(w*x + b)
  h2 = activation(w1*h1 + b1)
  return activation(w2*h2 + b2)

y_hat1 = feed_forward(x1)
print(y_hat1)
y_hat2 = feed_forward(x2)
print(y_hat2)
y_hat3 = feed_forward(x3)
print(y_hat3)
```

# Backpropagation

The backprop algorithm is derived from the goal of minimizing the error (or loss) function $\epsilon = (\hat{y} -y)^2$.

***
**Backpropagation for the current case:**
***
**2nd Hidden Layer**

$y_{hat} = activation(w_2*h_2 + b_2)$

$\frac {\partial{\epsilon}}{\partial{b_2}} = \frac {\partial{(y_{hat} - y)^2}}{\partial{b_2}} = 2 * (y_{hat} - y) * \frac {\partial{(y_{hat} - y)}}{\partial{b_2}} = 2 * (y_{hat} - y) * \frac {\partial{(activation(w_2*h_2 + b_2))}}{\partial{(w_2*h_2 + b_2)}} * \frac {\partial{(w_2*h_2 + b_2)}}{\partial{b_2}} = 2 * (y_{hat} - y) * \frac {\partial{(activation(w_2*h_2 + b_2))}}{\partial{(w_2*h_2 + b_2)}} * 1$

$\frac {\partial{\epsilon}}{\partial{w_2}} = \frac {\partial{(y_{hat} - y)^2}}{\partial{w_2}} = 2 * (y_{hat} - y) * \frac {\partial{(y_{hat} - y)}}{\partial{w_2}} = 2 * (y_{hat} - y) * \frac {\partial{(activation(w_2*h_2 + b_2))}}{\partial{(w_2*h_2 + b_2)}} * \frac {\partial{(w_2*h_2 + b_2)}}{\partial{w_2}} = 2 * (y_{hat} - y) * \frac {\partial{(activation(w_2*h_2 + b_2))}}{\partial{(w_2*h_2 + b_2)}} * h_2$

$\frac {\partial{\epsilon}}{\partial{h_2}} = \frac {\partial{(y_{hat} - y)^2}}{\partial{h_2}} = 2 * (y_{hat} - y) * \frac {\partial{(y_{hat} - y)}}{\partial{h_2}} = 2 * (y_{hat} - y) * \frac {\partial{(activation(w_2*h_2 + b_2))}}{\partial{(w_2*h_2 + b_2)}} * \frac {\partial{(w_2*h_2 + b_2)}}{\partial{h_2}} = 2 * (y_{hat} - y) * \frac {\partial{(activation(w_2*h_2 + b_2))}}{\partial{(w_2*h_2 + b_2)}} * w_2$

***
**1st Hidden Layer**

$h_2 = activation(w_1*h_1 + b_1)$

$\frac {\partial{\epsilon}}{\partial{b_1}} = \frac {\partial{\epsilon}}{\partial{h_2}} * \frac {\partial{h_2}}{\partial{b_1}} = \frac {\partial{\epsilon}}{\partial{h_2}} * \frac {\partial{(activation(w_1*h_1 + b_1))}}{\partial{b_1}} = \frac {\partial{\epsilon}}{\partial{h_2}} * \frac {\partial{(activation(w_1*h_1 + b_1))}}{\partial{(w_1*h_1 + b_1)}} * 1$

$\frac {\partial{\epsilon}}{\partial{w_1}} = \frac {\partial{\epsilon}}{\partial{h_2}} * \frac {\partial{h_2}}{\partial{w_1}} = \frac {\partial{\epsilon}}{\partial{h_2}} * \frac {\partial{(activation(w_1*h_1 + b_1))}}{\partial{w_1}} = \frac {\partial{\epsilon}}{\partial{h_2}} * \frac {\partial{(activation(w_1*h_1 + b_1))}}{\partial{(w_1*h_1 + b_1)}} * h_1$

$\frac {\partial{\epsilon}}{\partial{h_1}} = \frac {\partial{\epsilon}}{\partial{h_2}} * \frac {\partial{h_2}}{\partial{h_1}} = \frac {\partial{\epsilon}}{\partial{h_2}} * \frac {\partial{(activation(w_1*h_1 + b_1))}}{\partial{h_1}} = \frac {\partial{\epsilon}}{\partial{h_2}} * \frac {\partial{(activation(w_1*h_1 + b_1))}}{\partial{(w_1*h_1 + b_1)}} * w_1$

***
**Input Layer**

$h_1 = activation(w*x + b)$

$\frac {\partial{\epsilon}}{\partial{b}} = \frac {\partial{\epsilon}}{\partial{h_1}} * \frac {\partial{h_1}}{\partial{b}} = \frac {\partial{\epsilon}}{\partial{h_1}} * \frac {\partial{(activation(w*x + b))}}{\partial{b}} = \frac {\partial{\epsilon}}{\partial{h_1}} * \frac {\partial{(activation(w*x + b))}}{\partial{(w*x + b)}} * 1$

$\frac {\partial{\epsilon}}{\partial{w}} = \frac {\partial{\epsilon}}{\partial{h_1}} * \frac {\partial{h_1}}{\partial{w}} = \frac {\partial{\epsilon}}{\partial{h_1}} * \frac {\partial{(activation(w*x + b))}}{\partial{w}} = \frac {\partial{\epsilon}}{\partial{h_1}} * \frac {\partial{(activation(w*x + b))}}{\partial{(w*x + b)}} * x$
***

```
def derivative_activation(x):
  return 0 if x <= 0 else 1

def feed_forward_and_back_propagate(x, y, weights=np.array([b, w, b1, w1, b2, w2]), epochs=100, alpha=0.1):
  itr = 0
  while True:
    # forward propogate and predict y (and each unit in the graph)
    h1 = activation(weights[1]*x + weights[0])
    h2 = activation(weights[3]*h1 + weights[2])
    y_hat = activation(weights[5]*h2 + weights[4])    
    
    if itr >= epochs:
      break
    
    itr += 1
    
    # backpropagation to calculate gradients (for gradient descent)
    
    # gradients w.r.t 2nd hidden layer => activation(w2*h2 + b2)
    gradient_b2 = 2 * (y_hat-y) * derivative_activation(y_hat)
    gradient_w2 = 2 * (y_hat-y) * derivative_activation(y_hat) * h2
    gradient_h2 = 2 * (y_hat-y) * derivative_activation(y_hat) * w2
    
    # gradients w.r.t 1st hidden layer => activation(w1*h1 + b1)
    gradient_b1 = gradient_h2 * derivative_activation(h2)
    gradient_w1 = gradient_h2 * derivative_activation(h2) * h1
    gradient_h1 = gradient_h2 * derivative_activation(h2) * w1
    
    # gradients w.r.t input layer => activation(w*x + b)
    gradient_b = gradient_h1 * derivative_activation(h1)
    gradient_w = gradient_h1 * derivative_activation(h1) * x
    
    # update weights using gradient desecent => weights = weights - alpha*weight_gradients
    weight_gradients = [gradient_b, gradient_w, gradient_b1, gradient_w1, gradient_b2, gradient_w2]
    for i, gradient in enumerate(weight_gradients):
      weights[i] = weights[i] - alpha*gradient

  return y_hat
```

```
y1_hat = feed_forward_and_back_propagate(x1, y1)
print(f'y1 = {y1}, y1_hat = {np.round(y1_hat, 1)}')

y2_hat = feed_forward_and_back_propagate(x2, y2)
print(f'y2 = {y2}, y2_hat = {np.round(y2_hat, 1)}')

y3_hat = feed_forward_and_back_propagate(x3, y3)
print(f'y3 = {y3}, y3_hat = {np.round(y3_hat, 1)}')
```
