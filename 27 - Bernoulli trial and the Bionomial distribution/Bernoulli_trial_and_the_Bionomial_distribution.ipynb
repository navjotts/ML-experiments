# Overview

In a nutshell, a Bernoulli trial is an random experiment with 2 possible outcomes i.e. Success (Yes) or Failure (No).

https://en.wikipedia.org/wiki/Bernoulli_trial

## Problem Statement

Let's say the DCVC venture fund invests in 100 startups. Anywhere between 0 and 100 of the startups will succeed.

Construct a biomial distribution of getting a given number of startup successes, given that the probability of a successful startup is p = 0.1


# Experiment

```
import numpy as np

def conduct_binomial_trails(n, p):
  number_successful = 0
  for i in range(0, n):
    if np.random.random() < p: number_successful += 1
  return number_successful
```

```
NUMBER_OF_TRIALS = 100
PROBABILITY_OF_SUCCESS = 0.1

number_of_successes = conduct_binomial_trails(NUMBER_OF_TRIALS, PROBABILITY_OF_SUCCESS)
number_of_successes
```

**We want to run the above experiment many times – to construct the distribution we desire **

```
TOTAL = 10000

b_array = []
for i in range(0, TOTAL):
  b_array.append(conduct_binomial_trails(NUMBER_OF_TRIALS, PROBABILITY_OF_SUCCESS))
```

# Visualize

Setting `normed` to true in the `hist()` will esnure that the counts in the array are normalized to form a probability density

```
import matplotlib.pyplot as plt

plt.hist(b_array, normed=True)
plt.xlabel("Number of startup successes")
plt.ylabel("Probability")
plt.title("Bionomial Distribution for Startup Success Rate");
```
