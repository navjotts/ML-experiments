# Meshgrid

**The purpose of `meshgrid` is to create a rectangular grid out of an array of x_values and an array of y_values**

Example:

```
import numpy as np

x_values = np.array([1, 2, 3, 4, 5])
yvalues = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90])

xx, yy = np.meshgrid(x_values, yvalues)
print(xx)
print(yy)
```

# The Mexican Hat Filter or Laplacian of Gaussian

```
def laplacian_of_gaussian(x,y):
    sigma = 0.3
    return 1/np.pi*sigma * (1-0.5*((x**2 + y**2)/sigma**2)) * np.exp(-((x**2+y**2)/(2*sigma**2)))
```

**3D plot using linspace without meshgrid**

```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d

fig, ax = plt.subplots()
fig.set_figwidth(12)
fig.set_figheight(5.5/8*12)

ax = plt.axes(projection='3d')

x = np.linspace(-1, 1, 1000)
y = np.linspace(-1, 1, 1000)
z = laplacian_of_gaussian(x, y)

ax.plot3D(x, y, z, '-', color='black', label='z = laplacian_of_gaussian(x,y)');
ax.legend(prop={'family': 'serif', 'size': 12})

fig.suptitle('The Mexican Hat Filter or Laplacian of Gaussian', fontdict={'family': 'serif', 'size': 18}, y=0.92);
```

**3D plot using a random distribution without meshgrid **

```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d

fig, ax = plt.subplots()
fig.set_figwidth(12)
fig.set_figheight(5.5/8*12)

ax = plt.axes(projection='3d')

x = np.random.uniform(-1, 1, 1000)
y = np.random.uniform(-1, 1, 1000)
z = laplacian_of_gaussian(x, y)

ax.plot3D(x, y, z, 'o', color='black', label='z = laplacian_of_gaussian(x,y)');
ax.legend(prop={'family': 'serif', 'size': 12})

fig.suptitle('The Mexican Hat Filter or Laplacian of Gaussian', fontdict={'family': 'serif', 'size': 18}, y=0.92);
```

**3D plot using meshrgrid**

```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d

fig, ax = plt.subplots()
fig.set_figwidth(12)
fig.set_figheight(5.5/8*12)

ax = plt.axes(projection='3d')

x = np.linspace(-1, 1, 100)
y = np.linspace(-1, 1, 100)
X, Y = np.meshgrid(x, y)
Z = laplacian_of_gaussian(X, Y)

ax.contour3D(X, Y, Z, 200, cmap='RdYlGn')
ax.view_init(60, 35)

fig.suptitle('The Mexican Hat Filter or Laplacian of Gaussian', fontdict={'family': 'serif', 'size': 18}, y=0.92);
```

```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d

fig, ax = plt.subplots()
fig.set_figwidth(12)
fig.set_figheight(5.5/8*12)

ax = plt.axes(projection='3d')

x = np.linspace(-1, 1, 100)
y = np.linspace(-1, 1, 100)
X, Y = np.meshgrid(x, y)
Z = laplacian_of_gaussian(X, Y)

ax.plot_surface(X, Y, Z, rstride=1, cstride=1, cmap='gist_heat', edgecolor=None)
ax.set(xlabel='x', ylabel='y', zlabel='z = laplacian_of_gaussian(x,y)', title='The Mexican Hat Filter or Laplacian of Gaussian');
```
